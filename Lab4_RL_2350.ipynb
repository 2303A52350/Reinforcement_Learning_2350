```python
import math


machines = ['A', 'B', 'C']


N_play = 10


N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}
UCB = {'A': 0.0, 'B': 0.0, 'C': 0.0}

total_reward = 0

print("=== Upper Confidence Bound (UCB) Multi-Armed Bandit ===")
print("Machines:", machines)
print("Number of plays:", N_play)


for play in range(1, N_play + 1):


    for machine in machines:
        if N[machine] == 0:
            UCB[machine] = float('inf')
        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(play)) / N[machine]
            )


    selected_machine = max(machines, key=lambda x: UCB[x])

    print("\n--------------------------------")
    print("Play", play)
    print("UCB values:")

    for machine in machines:
        if UCB[machine] == float('inf'):
            print(machine, " = Infinity")
        else:
            print(machine, " = {:.4f}".format(UCB[machine]))

    print("Play with machine", selected_machine)


    reward = float(input("Enter reward: "))


    N[selected_machine] += 1
    R[selected_machine] += reward


    Q[selected_machine] = Q[selected_machine] + (
        reward - Q[selected_machine]
    ) / N[selected_machine]

    total_reward += reward

    print("Updated Q-value of", selected_machine,
          "=", round(Q[selected_machine], 4))


print("\n\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tR\tUCB\t\tQ")


for machine in machines:
    if N[machine] == 0:
        UCB[machine] = float('inf')
    else:
        UCB[machine] = Q[machine] + math.sqrt(
            (2 * math.log(N_play)) / N[machine]
        )

    print("{}\t{}\t{:.1f}\t{:.4f}\t\t{:.4f}".format(
        machine,
        N[machine],
        R[machine],
        UCB[machine],
        Q[machine]
    ))

print("\nTotal Reward =", total_reward)
```
