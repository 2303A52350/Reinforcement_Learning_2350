```python
import random

machines = ['A', 'B', 'C']

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

epsilon = 0.2
num_plays = 10
total_reward = 0

print("===== Epsilon-Greedy Multi-Armed Bandit =====")
print("Machines:", machines)
print("Epsilon:", epsilon)
print("Number of Plays:", num_plays)

for play in range(1, num_plays + 1):

    r = random.random()

    if r < epsilon:
        machine = random.choice(machines)
        decision = "Exploration"
    else:

        max_q = max(Q.values())
        best_machines = [m for m in machines if Q[m] == max_q]
        machine = random.choice(best_machines)
        decision = "Exploitation"

    print(f"\nPlay {play}")
    print(f"Random number = {r:.4f}")
    print(f"{decision}: Machine {machine}")

    reward = float(input(f"Enter reward for Machine {machine}: "))

    N[machine] += 1
    R[machine] += reward

    Q[machine] = Q[machine] + (reward - Q[machine]) / N[machine]

    total_reward += reward

    print(f"N({machine}) = {N[machine]}")
    print(f"R({machine}) = {R[machine]:.2f}")
    print(f"Q({machine}) = {Q[machine]:.2f}")

print("\n========== FINAL RESULTS ==========")

print("\nMachine\tN(a)\tR(a)\tQ(a)")

for machine in machines:
    print(f"{machine}\t{N[machine]}\t{R[machine]:.2f}\t{Q[machine]:.2f}")

print("\nTotal Reward after 10 plays =", total_reward)

best_machine = max(Q, key=Q.get)
print("Best Estimated Machine =", best_machine)
print("Best Q-value =", round(Q[best_machine], 2))
```
