```python
import random


machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}


print("Modified Epsilon-Greedy Multi-Armed Bandit")
print("-------------------------------------------")


for play in range(1, N_play + 1):

    print(f"\nPlay {play}")

    r = random.random()

    max_q = max(Q.values())


    best_machines = [m for m in machines if Q[m] == max_q]
    best_machine = random.choice(best_machines)

    if r < epsilon:
        other_machines = [m for m in machines if m != best_machine]
        selected_machine = random.choice(other_machines)

        print("Exploration")
    else:
        selected_machine = best_machine

        print("Exploitation")

    print("Play with machine", selected_machine)

    reward = float(input("Enter reward: "))

    N[selected_machine] += 1

    R[selected_machine] += reward

    Q[selected_machine] = Q[selected_machine] + (
        reward - Q[selected_machine]
    ) / N[selected_machine]

    print("Updated Q-value =", Q[selected_machine])


print("\n\nFINAL RESULTS")
print("-------------------------------------------")
print("Machine\tN\tR\tQ")

for machine in machines:
    print(f"{machine}\t{N[machine]}\t{R[machine]}\t{Q[machine]:.2f}")


total_reward = sum(R.values())

print("\nTotal Reward Received =", total_reward)
```
