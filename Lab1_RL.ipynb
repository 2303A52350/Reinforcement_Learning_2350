```python
#Question1
n = int(input("Enter number of machines: "))

machines = {}
q_values = {}

for i in range(n):
    name = input(f"Enter machine name {i + 1}: ")
    rewards = list(map(int, input("Enter rewards separated by spaces: ").split()))

    machines[name] = rewards
    q_values[name] = sum(rewards) / len(rewards)

print("\nEstimated Q(a) values:")

for name, rewards in machines.items():
    print(f"Machine {name}")
    print(f"Rewards: {rewards}")
    print(f"Q({name}) = {q_values[name]:.2f}")

selected = max(q_values, key=q_values.get)

print(f"\nGreedy Selected Machine: {selected}")
```

```python

machines = {
    "A": [3, 2, 4],
    "B": [6, 7],
    "C": [5, 4, 6]
}

Q = {"A": 0.0, "B": 0.0, "C": 0.0}
N = {"A": 0, "B": 0, "C": 0}

print("Incremental Update Process\n")

for machine, rewards in machines.items():
    for reward in rewards:
        N[machine] += 1

        Q[machine] = Q[machine] + (reward - Q[machine]) / N[machine]

        print(f"Machine {machine}, Reward = {reward}")
        print(f"N({machine}) = {N[machine]}")
        print(f"Q({machine}) = {Q[machine]:.2f}")

        selected = max(Q, key=Q.get)
        print(f"Current Greedy Selection: Machine {selected}")
        print("-" * 35)

print("\nFinal Results:")

for machine in machines:
    print(f"Machine {machine}")
    print(f"N({machine}) = {N[machine]}")
    print(f"Q({machine}) = {Q[machine]:.2f}")

selected = max(Q, key=Q.get)
print(f"\nGreedy Selected Machine: {selected}")
```
